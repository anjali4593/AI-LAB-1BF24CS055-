```python
import pandas as pd
import itertools
import re
from IPython.display import display

# Evaluate propositional logic formulas
def eval_formula(formula, env):
    f = formula.strip()

    # Convert logical implication: P -> Q
    f = f.replace("->", "<=")

    # Convert negation: ~Q -> (not Q)
    f = re.sub(r"~\s*([A-Za-z]\w*)", r"(not \1)", f)

    # Convert OR and AND
    f = re.sub(r"\s+v\s+", " or ", f)
    f = f.replace("^", " and ")

    return bool(eval(f, {"__builtins__": {}}, env))


# Propositional variables
variables = ["P", "Q", "R"]

# Knowledge Base
kb_formulas = [
    "Q -> P",
    "P -> ~Q",
    "Q v R"
]

# Queries to test
queries = ["R", "R -> P", "Q -> R"]

# Generate all truth assignments
models = list(itertools.product([True, False], repeat=3))
rows = []

for model in models:
    env = dict(zip(variables, model))

    # Evaluate each sentence in the KB
    kb_values = {
        formula: eval_formula(formula, env)
        for formula in kb_formulas
    }

    # KB is true only when all sentences are true
    kb_true = all(kb_values.values())

    # Create the table row
    row = {
        "P": env["P"],
        "Q": env["Q"],
        "R": env["R"],
        "Q => P": kb_values["Q -> P"],
        "P => ~Q": kb_values["P -> ~Q"],
        "Q v R": kb_values["Q v R"],
        "KB": kb_true
    }

    # Show query values only for models satisfying KB
    for query in queries:
        if kb_true:
            row[f"KB entails {query}"] = eval_formula(query, env)
        else:
            row[f"KB entails {query}"] = "N/A"

    rows.append(row)

# Display truth table
df = pd.DataFrame(rows)
display(df)

# Check entailment across every model of the KB
kb_models = [row for row in rows if row["KB"]]

print("\n--- Entailment Results ---")

for query in queries:
    if not kb_models:
        result = True  # Vacuous entailment for inconsistent KB
    else:
        result = all(
            eval_formula(query, {
                "P": row["P"],
                "Q": row["Q"],
                "R": row["R"]
            })
            for row in kb_models
        )

    print(f"Does KB entail '{query}'? {result}")
```
