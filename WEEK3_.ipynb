```python
import heapq

# Goal state
GOAL = (1, 2, 3,
        8, 0, 4,
        7, 6, 5)

# Possible movements of blank (0)
MOVES = [(-1, 0), (1, 0), (0, -1), (0, 1)]


def misplaced_tiles(state):
    """Heuristic: Number of tiles not in their goal position."""
    count = 0

    for i in range(9):
        if state[i] != 0 and state[i] != GOAL[i]:
            count += 1

    return count


def get_neighbors(state):
    """Generate all possible next states."""
    neighbors = []

    zero = state.index(0)
    row, col = divmod(zero, 3)

    for dr, dc in MOVES:
        new_row = row + dr
        new_col = col + dc

        if 0 <= new_row < 3 and 0 <= new_col < 3:
            new_zero = new_row * 3 + new_col

            new_state = list(state)
            new_state[zero], new_state[new_zero] = \
                new_state[new_zero], new_state[zero]

            neighbors.append(tuple(new_state))

    return neighbors


def print_board(state):
    for i in range(0, 9, 3):
        print(state[i:i+3])
    print()


def a_star_misplaced(start):
    # Priority queue: (f, g, state, path)
    pq = []

    h = misplaced_tiles(start)
    heapq.heappush(pq, (h, 0, start, [start]))

    visited = set()

    while pq:
        f, g, state, path = heapq.heappop(pq)

        if state in visited:
            continue

        visited.add(state)

        if state == GOAL:
            return path

        for neighbor in get_neighbors(state):
            if neighbor not in visited:
                new_g = g + 1
                new_h = misplaced_tiles(neighbor)
                new_f = new_g + new_h

                heapq.heappush(
                    pq,
                    (new_f, new_g, neighbor, path + [neighbor])
                )

    return None


# Initial state
start = (2, 8, 3,
         1, 6, 4,
         7, 0, 5)

solution = a_star_misplaced(start)

print("A* using Misplaced Tiles Heuristic")
print("Number of moves:", len(solution) - 1)

for state in solution:
    print_board(state)
```

```python
import heapq

# Goal state
GOAL = (1, 2, 3,
        8, 0, 4,
        7, 6, 5)

# Possible movements of blank
MOVES = [(-1, 0), (1, 0), (0, -1), (0, 1)]


def manhattan_distance(state):
    """Heuristic: Sum of Manhattan distances of all tiles."""
    distance = 0

    for i in range(9):
        tile = state[i]

        # Don't calculate distance for blank
        if tile == 0:
            continue

        # Current position
        current_row, current_col = divmod(i, 3)

        # Goal position
        goal_index = GOAL.index(tile)
        goal_row, goal_col = divmod(goal_index, 3)

        # Manhattan distance
        distance += abs(current_row - goal_row) + \
                    abs(current_col - goal_col)

    return distance


def get_neighbors(state):
    """Generate all possible next states."""
    neighbors = []

    zero = state.index(0)
    row, col = divmod(zero, 3)

    for dr, dc in MOVES:
        new_row = row + dr
        new_col = col + dc

        if 0 <= new_row < 3 and 0 <= new_col < 3:
            new_zero = new_row * 3 + new_col

            new_state = list(state)
            new_state[zero], new_state[new_zero] = \
                new_state[new_zero], new_state[zero]

            neighbors.append(tuple(new_state))

    return neighbors


def print_board(state):
    for i in range(0, 9, 3):
        print(state[i:i+3])
    print()


def a_star_manhattan(start):
    # Priority queue: (f, g, state, path)
    pq = []

    h = manhattan_distance(start)
    heapq.heappush(pq, (h, 0, start, [start]))

    visited = set()

    while pq:
        f, g, state, path = heapq.heappop(pq)

        if state in visited:
            continue

        visited.add(state)

        if state == GOAL:
            return path

        for neighbor in get_neighbors(state):
            if neighbor not in visited:
                new_g = g + 1
                new_h = manhattan_distance(neighbor)
                new_f = new_g + new_h

                heapq.heappush(
                    pq,
                    (new_f, new_g, neighbor, path + [neighbor])
                )

    return None


# Initial state
start = (2, 8, 3,
         1, 6, 4,
         7, 0, 5)

solution = a_star_manhattan(start)

print("A* using Manhattan Distance Heuristic")
print("Number of moves:", len(solution) - 1)

for state in solution:
    print_board(state)
```
